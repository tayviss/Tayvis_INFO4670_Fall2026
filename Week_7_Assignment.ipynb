```python
import warnings
warnings.filterwarnings('ignore', category=DeprecationWarning)
!pip install mlxtend networkx

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from mlxtend.frequent_patterns import fpgrowth, association_rules

df = pd.read_csv('bread_basket.csv')
print("Dataset successfully loaded! Shape:", df.shape)
print(df.head())
```

Section 2

```python
print(df.dtypes)
```

```python
print(df.describe(include='all'))
```

```python
subtitle = "Tayvis Sanders"
item_counts = df.groupby('item')['transaction'].nunique().sort_values(ascending=False)

ax = item_counts.head(15).plot(kind='bar', figsize=(12, 5))
plt.title(f"Transaction Count per Item\nSubtitle: {subtitle}")
plt.xlabel("Item")
plt.ylabel("Unique Transactions")
plt.tight_layout()
plt.show()
```

```python
target_items = ['Coffee', 'Tea', 'Alfajores', 'Juice', 'Chicken Stew']
counts = df[df['item'].isin(target_items)]['item'].value_counts()
print(counts)
```

Section 3

```python
basket = df.groupby(['transaction', 'item'])['item'].count().unstack().fillna(0)
basket = basket.applymap(lambda x: 1 if x > 0 else 0).astype(bool)

print("One-hot basket shape (transactions x items):", basket.shape)

frequent_itemsets = fpgrowth(basket, min_support=0.02, use_colnames=True)

print(f"\nTotal frequent itemsets found: {len(frequent_itemsets)}")
print(frequent_itemsets.head(15))
```

Section 4

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.2)

rules = rules.sort_values(by="lift", ascending=False)

report_table = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
print(f"Total rules generated: {len(rules)}")
print(report_table.head(15))
```

Section 5

```python
core_items = {'Coffee', 'Bread', 'Cake', 'Tea'}

filtered_rules = rules[
    rules['antecedents'].apply(lambda x: any(item in core_items for item in x)) &
    rules['consequents'].apply(lambda x: any(item in core_items for item in x))
]

fig, ax = plt.subplots(figsize=(10, 6))
G = nx.DiGraph()

for idx, row in filtered_rules.head(15).iterrows():
    ant = ', '.join(list(row['antecedents']))
    cons = ', '.join(list(row['consequents']))
    G.add_edge(ant, cons, weight=row['lift'])

pos = nx.spring_layout(G, k=0.6, seed=42)
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=3000, font_size=9, width=2, arrowsize=20)
plt.title("Association Rule Subgraph (Bread, Coffee, Cake, Tea)")
plt.show()
```

My interpretation of {Coffee, Cake} -> {Bread}:

- Support: Represents the fraction of all
transactions in the dataset that contain coffee, cake, and bread simultaneously. It measures how frequently this combination occurs overall across all customer baskets.

- Confidence: Represents the conditional probability, among all customer baskets containing both coffee and cake, the exact percentage that also include bread.

- Lift > 1: Indicates a positive correlation between purchasing coffee & cake and purchasing bread. A lift greater than 1 means that customers who buy coffee and cake are significantly more likely to buy bread than expected by random chance, showing a strong complementary purchasing behavior useful for cross-merchandising or bundle promotions.
